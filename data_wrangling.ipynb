```python
# Q1
students = pd.read_csv('students_master.csv')
marks = pd.read_csv('exam_marks.csv')
print(students.head())
print(marks.head())
print("Common column:", list(set(students.columns) & set(marks.columns)))
```

```python
# Q2
inner = pd.merge(students, marks, on='Student_ID')
print(inner)
```

```python
# Q3
left = pd.merge(students, marks, on='Student_ID', how='left')
print(left)
```

```python
# Q4
right = pd.merge(students, marks, on='Student_ID', how='right')
print(right)
```

```python
# Q5
outer = pd.merge(students, marks, on='Student_ID', how='outer')
print(outer)
print(outer['Student_ID'].tolist())
```

```python
# Q6
ca = pd.read_csv('section_CA.csv')
cb = pd.read_csv('section_CB.csv')
combined = pd.concat([ca, cb])
print(combined)
```

```python
# Q7
combined2 = pd.concat([ca, cb], ignore_index=True)
print(combined2)
print(combined2.index)
```

```python
# Q8
side_by_side = pd.concat([ca, cb], axis=1)
print(side_by_side)
print(side_by_side.columns)
```

```python
# Q9
s1 = ca['Attendance']
s2 = cb['Student_ID']
print(pd.concat([s1, s2], axis=0))
```

```python
# Q10
orig = pd.read_csv('marks_original.csv')
upd = pd.read_csv('marks_updated.csv')
print(orig[orig['Marks'].isnull()])
print("Missing in original:", orig['Marks'].isnull().sum())
print(upd[upd['Marks'].isnull()])
print("Missing in updated:", upd['Marks'].isnull().sum())
```

```python
# Q11
orig_i = orig.set_index('Student_ID')
upd_i = upd.set_index('Student_ID')
print(orig_i.combine_first(upd_i))
```

```python
# Q12
print(orig_i.combine(upd_i, np.maximum))
```

```python
# Q13
print(orig_i.combine(upd_i, np.fmax))
print("np.maximum gives NaN whenever one of the two marks is missing, but np.fmax ignores the NaN and returns the available mark. So Student 102, 105, 107 and 108 get a mark with fmax, but NaN with maximum.")
```

```python
# Q14
long_df = pd.read_csv('marks_long.csv')
wide = long_df.pivot(index='Student_ID', columns='Subject', values='Marks')
print(wide)
```

```python
# Q15
stacked = wide.stack()
print(stacked)
```

```python
# Q16
enroll = pd.read_csv('semester_enrollments.csv')
enr_inner = pd.merge(enroll, students, on='Student_ID', how='inner')
print(enr_inner)
print("Some students appear in more than one row because they are enrolled in more than one semester/course (101, 102 and 105 have two enrollment records). Each enrollment row is matched with the single master row of that student, so the student details repeat.")
```

```python
# Q17
enr_left = pd.merge(enroll, students, on='Student_ID', how='left')
print(enr_left)
print(enr_left[enr_left['Student_Name'].isnull()])
```

```python
# Q18
joins = {}
for how in ['inner', 'left', 'right', 'outer']:
    joins[how] = len(pd.merge(students, marks, on='Student_ID', how=how))
print(pd.Series(joins, name='rows'))
print("inner keeps only matching IDs, left keeps all 8 students, right keeps all 7 marks records and outer keeps every ID from both files, so outer has the most rows.")
```

```python
# Q19
default_concat = pd.concat([ca, cb])
print(default_concat)
print(default_concat.isnull().sum())
print("Attendance and Club have missing values because Attendance exists only in section_CA and Club exists only in section_CB. With the default join='outer' all columns are kept and the missing ones are filled with NaN.")
```

```python
# Q20
inner_concat = pd.concat([ca, cb], join='inner', ignore_index=True)
print(inner_concat)
print("Remaining columns:", list(inner_concat.columns))
```

```python
# Q21
orig2 = pd.read_csv('marks_original.csv')
upd2 = pd.read_csv('marks_updated.csv')
print(orig2.combine_first(upd2))
print(orig2.set_index('Student_ID').combine_first(upd2.set_index('Student_ID')))
print("Without the index, combine_first matches rows by their position (0, 1, 2 ...). Since the updated file is in a different order, the marks of the wrong students get filled (for example Student 102 gets 90, which actually belongs to 104, and Student 108 stays missing). After setting Student_ID as the index the alignment is done by ID, so the correct one is the indexed version.")
```

```python
# Q22
final_marks = orig_i.combine(upd_i, np.fmax).sort_index()
print(final_marks)
```

```python
# Q23
unstacked = stacked.unstack()
print(unstacked)
print("Matches the pivot table:", unstacked.equals(wide))
```

```python
# Q24
repeat = pd.read_csv('repeat_tests.csv')
try:
    repeat.pivot(index='Student_ID', columns='Subject', values='Marks')
except ValueError as err:
    print("Error:", err)
print("pivot() cannot work here because some Student_ID and Subject combinations are repeated (for example 101-Python, 103-Python and 104-DBMS have two marks). pivot() does not know which value to put in that one cell and it does not aggregate.")
```

```python
# Q25
print(repeat.pivot_table(index='Student_ID', columns='Subject', values='Marks', aggfunc='mean'))
```

```python
# Q26
roster = pd.concat([ca, cb], ignore_index=True)
roster_marks = pd.merge(roster, marks, on='Student_ID', how='left')
result = roster_marks[['Student_ID', 'Student_Name', 'Section', 'Python_Marks', 'DBMS_Marks']]
print(result)
print(result[result['Python_Marks'].isnull() | result['DBMS_Marks'].isnull()])
```

```python
# Q27
best = orig_i.combine_first(upd_i)
final = pd.merge(students, best.reset_index(), on='Student_ID', how='left')
print(final)
print(orig_i.loc[[101, 102]])
print(upd_i.loc[[101, 102]])
print("Student 101: the original mark is 85 (not missing), so combine_first keeps 85 and the updated mark 87 is not used. Student 102: the original mark is missing, so it is filled from the updated file and the final value is 89.")
```

```python
# Q28
a = orig_i.combine_first(upd_i)
b = orig_i.combine(upd_i, np.fmax)
compare = pd.DataFrame({'combine_first': a['Marks'], 'combine_fmax': b['Marks']})
print(compare)
diff = compare[compare['combine_first'] != compare['combine_fmax']]
print(diff)
print("IDs where they differ:", diff.index.tolist())
```

```python
# Q29
pt = repeat.pivot_table(index='Student_ID', columns='Subject', values='Marks', aggfunc=['mean', 'max'])
print(pt)
stk = pt.stack()
print(stk)
restored = stk.unstack()
print(restored)
print("Same as original pivot table:", restored.reindex(columns=pt.columns).equals(pt))
```

```python
# Q30
merged = pd.merge(students, long_df, on='Student_ID')
wide2 = merged.pivot(index=['Student_ID', 'Student_Name'], columns='Subject', values='Marks')
print(wide2)
wide2 = wide2.reset_index()
print(wide2)
print("This pivot() works because in marks_long every Student_ID and Subject combination appears only once, so each cell has exactly one value. In repeat_tests some combinations are repeated, so pivot() gets more than one value for the same cell and raises an error. pivot_table() with an aggfunc is needed there.")
```
